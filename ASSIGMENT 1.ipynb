### 8. Student Eligibility Create variables for marks and attendance. A student is eligible if: • Marks are greater than or equal to 50 • Attendance is greater than or equal to 75 Use logical operators to calculate and print whether the student is eligible. ###

```python
attendance = int(input("enter your attendance"))
marks = int(input("enter your marks"))

min_attendance = 75
min_marks = 50

ok = attendance >= min_attendance and marks >= min_marks
print(f"eligible {ok} ")
```
